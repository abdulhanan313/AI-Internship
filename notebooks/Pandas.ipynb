Data frames operations

```python
import pandas as pd

# 1. Load the original dataset
df = pd.read_csv("ecommerce_sales_data.csv")
Year	Month	Category	Region	Units_Sold	Revenue	Profit	Customer_Rating	Ad_Spend


# Define the target columns explicitly
cost_columns = ["Cat_id", "Category", "Units_for_sale", "Cost"]
revenue_columns = ["Cat_id", "Category", "Units_sold", "Revenue"]

# Function to extract columns and fill missing ones with NaN
def extract_and_ensure_columns(dataframe, target_cols, filename):
    # Copy only the columns that actually exist in the dataframe
    existing_cols = [col for col in target_cols if col in dataframe.columns]
    extracted_df = dataframe[existing_cols].copy()
    
    # Add any missing target columns and fill them with empty values (NaN)
    for col in target_cols:
        if col not in extracted_df.columns:
            extracted_df[col] = None  # Or use pd.NA / float('nan')
            
    # Rearrange columns to match the exact target order and remove duplicates
    final_df = extracted_df[target_cols].drop_duplicates()
    
    # Save the complete file
    final_df.to_csv(filename, index=False)
    print(f"Saved: {filename} with shape {final_df.shape}")

# 2. Generate both complete CSV files from the loaded data
extract_and_ensure_columns(df, cost_columns, "Products_cost.csv")
extract_and_ensure_columns(df, revenue_columns, "Products_revenue.csv")
```

```python
import pandas as pd
df_cost = pd.read_csv("Products_cost.csv")
df_revenue = pd.read_csv("Products_revenue.csv")
print(df_cost.head(5))  # Display the first few rows of the cost DataFrame
print(df_revenue.head(5))  # Display the first few rows of the revenue Data
print("====================After Merging====================")
df_products_merged = pd.merge(df_cost, df_revenue, on=["Cat_ID"])
print(df_products_merged)  # Display the merged DataFrame without the index)
print("------------------------------------------------------")
```

```python
print(pd.set_option("display.max_columns",None))  # Set the option to display all columns in the DataFrame
df_products_merged.to_csv("Merged_cost_revenue.csv",index = False)  # Save the merged DataFrame to a CSV file without the index
```
