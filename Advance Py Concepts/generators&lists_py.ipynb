```python
# List comprehension
squares = [x**2 for x in range(10)]
evens = [x for x in range(20) if x % 2 == 0]

print(squares)
print(evens)
```

```python
# Dict comprehension
word_lengths = {w: len(w) for w in ["cat", "elephant", "ox"]}
print(word_lengths)
```

```python
# Set comprehension (dedupes automatically)
unique_lengths = {len(w) for w in ["cat", "elephant", "ox", "dog"]}
print(unique_lengths)
```

```python
# Nested comprehension (flatten a matrix)
matrix = [[1, 2], [3, 4], [5, 6]]
flat = [num for row in matrix for num in row]
print(flat)
print(matrix)
```

The Iteration Protocol

```python
nums = [1, 2, 3]
it = iter(nums)        # calls nums.__iter__()
print(next(it))        # 1 — calls it.__next__()
print(next(it))        # 2
print(next(it))        # 3
#next(it)               # raises StopIteration
```

A for x in nums: loop is just sugar for this exact iter()/next()/StopIteration dance. Generators (next section) are simply a shortcut for writing your own iterator without manually defining a class with __iter__/__next__.

3. Generators — and when they beat a list

A generator function uses yield instead of return. Calling it doesn't run the code — it returns a generator object that produces values one at a time, on demand.

```python
#Generators Fn
def squares_gen(n):
    for x in range(n):
        yield x**2

g = squares_gen(5)
print(next(g))   # 0
print(next(g))   # 1
print(next(g))   #2
print(next(g))   #3
print(next(g))   #4
```

```python
#Generator Expression
squares_gen = (x**2 for x in range(5))
for x in squares_gen:
  print(x)
```

When a generator beats a list:

The full result set is huge (or unbounded) and you only need to process it once, item by item
You're chaining/filtering a stream (e.g. reading a giant log file line by line)
You don't need random access (data[500]) or len()
You want to start producing results before the whole computation finishes

When a list is still better: you need to iterate multiple times, need indexing, need len(), or the data is small anyway (a generator adds no benefit and some overhead for tiny data).

```python
#Flatten Array
def flatten(iterable):
    result = []
    for item in iterable:
        if isinstance(item, list):
            result.extend(flatten(item))
        elif item is not None:
            result.append(item)
    return result
```
